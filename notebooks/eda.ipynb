# Análise exploratória EDA

Objetivo: Diagnosticar a integridade da tabela orders sem fazer alterações (limpeza de valores nulls, etc)

```python
#importando bibliotecas
import pandas as pd
```

```python
#carregando a tabela orders
df_orders = pd.read_csv('../data/orders.csv')
#exibindo as primeiras linhas do dataframe para a leitura dos dados
df_orders.head()
```

## Parte 01 - Visão geral (orders) e análise do intervalo de datas (created_at)

```python
#quantidade total de linhas do dataframe:
df_orders.shape[0]
print(f"Quantidade total de linhas do dataframe: {df_orders.shape[0]}")
```

```python
#quantidade total de colunas do dataframe:
df_orders.shape[1]
print(f"Quantidade total de colunas do dataframe: {df_orders.shape[1]}")
```

```python
#análise do intervalo de datas da tabela created_at

#conferindo se a coluna created_at está no formato datetime e convertendo para datetime caso não esteja
df_orders['created_at'] = pd.to_datetime(df_orders['created_at'])

#buscando a data minima e máxima da coluna created_at
data_min = df_orders['created_at'].min()
data_max = df_orders['created_at'].max()

print(f"Período analisado: de {data_min} até {data_max}")
```

## Parte 02 - Análise das métricas da coluna de faturamento (total)

```python
#visualizando a distribuição da coluna total
df_orders['total'].describe()
```

```python
# cálculo das métricas descritivas básicas para a coluna total

# cálculo do valor mínimo, máximo e média da coluna total
total_min = df_orders['total'].min()
total_max = df_orders['total'].max()
total_mean = df_orders['total'].mean()

print(f"Esse é o valor mínimo: R$ {total_min:.2f}")
print(f"Esse é o valor máximo: R$ {total_max:.2f}")
print(f"Esse é o valor médio: R$ {total_mean:.2f}")
```

## Parte 03 - Diagnóstico e interpretação

### Perguntas para o entendimento do negócio:
- No registro do ERP(Planejamento de Recursos Empresariais) existem valores negativos ou nulos?
- Há presença de outlier em "total"?
- Há valores ausentes ou nulos?
### O que foi encontrado:
Pelo que eu observei, em relação a coluna "total", não há valores negativos e nem valores nulos. Se a coluna "total" tivesse valores negativos ou nulos significaria que essa tabela possívelmente está misturada com dados de outros eventos como devoluções das vendas, no entanto, como não foram encontrados esses valores, isso mostra que a tabela se trata apenas de vendas brutas.

Em relação a presença de outlier na tabela "total", há uma discrepância significativa entre o valor médio(R$ 28.704,99), o valor máximo(R$ 127.262,02) e o valor mínimo (R$ 32,62) uma diferença de R$ 98.557,03 entre o valor médio e o máximo(quase 4 vezes o valor da média). No entanto, como a empresa vende diversos tipos de equipamentos e peças menores para embarcações essa discrepância é esperada por conta do modelo de negócio, porém se faz nescessário avaliar esses valores mais altos pois eles estão puxando para que o valor da média fique muito alto. Em análises futuras seria interessante usar a mediana como métrica central pois garantiria mais precisão ao medir os valores.

Sobre a qualidade e integridade dos dados,se há presença de valores nulos ou ausentes nas tabelas críticas("id", "customer_id", "created_at" e "total") não foram encontrados e estão preenchidas corretamente. Somente foram encontrados valores nulos na tabela "salesperson_id" o que pode significar que não houve a intermediação de um vendedor físico ou uma falha no mapeamento do ERP que precisa ser investigada e corrigida.

### Conclusão sobre o dataset
A tabela orders possui uma boa qualidade em sua estrutura (sem chaves nulas ou valores negativos). Se mostrando extremamente Confiável para métricas globais de faturamento, mas é preciso um tratamento e segmentação prévios (como separar vendas por canal e tratar outliers) antes de ser utilizada em modelos preditivos ou tomadas de decisão estratégicas.
